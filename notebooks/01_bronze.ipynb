```python
bronze_selic_count = spark.table(
    "beanalytic_case.bronze.selic"
).count()

bronze_ipca_count = spark.table(
    "beanalytic_case.bronze.ipca"
).count()

print(f"Bronze SELIC: {bronze_selic_count}")
print(f"Bronze IPCA: {bronze_ipca_count}")
```

```python
bronze_selic_distinct = (
    spark.table("beanalytic_case.bronze.selic")
    .select("data_raw")
    .distinct()
    .count()
)

bronze_ipca_distinct = (
    spark.table("beanalytic_case.bronze.ipca")
    .select("data_raw")
    .distinct()
    .count()
)

print(
    f"SELIC rows: {bronze_selic_count} | "
    f"Distinct dates: {bronze_selic_distinct}"
)

print(
    f"IPCA rows: {bronze_ipca_count} | "
    f"Distinct dates: {bronze_ipca_distinct}"
)
```
