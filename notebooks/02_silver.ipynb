```python
selic_validation = spark.sql(
    """
    SELECT
        COUNT(*) AS total_rows,
        COUNT(DISTINCT reference_date)
            AS distinct_business_keys
    FROM beanalytic_case.silver.selic
    """
)

display(selic_validation)
```

```python
ipca_validation = spark.sql(
    """
    SELECT
        COUNT(*) AS total_rows,
        COUNT(DISTINCT reference_month)
            AS distinct_business_keys
    FROM beanalytic_case.silver.ipca
    """
)

display(ipca_validation)
```
